```python
def multiply(multiplier):
    def multiply_generator(old_function):
        def new_function(*args, **kwds):
            return multiplier * old_function(*args, **kwds)
        return new_function
    return multiply_generator # it returns the new generator

# Usage
@multiply(3) # multiply is not a generator, but multiply(3) is
def return_num(num):
    return num

# Now return_num is decorated and reassigned into itself
return_num(5) # should return 15
```

# data collections

类型	英文名	   语法示例	   有序性	可变性	  重复元素	          主要特点		
列表	list	[1, 2, "a"]	 有序	✅ 可变	 ✅ 允许	    最常用的动态数组，支持任意类型嵌套。		
元组	tuple	(1, 2, "a")	 有序	🔒 不可变  ✅ 允许	“只读列表”，速度略快，内存占用小，可做字典键。		
字典	dict	{'k': 'v'}	 有序*	✅ 可变	 ❌ 键唯一	键值对映射，查找速度极快 (Hash Table)。		
集合	set	    {1, 2, 3}	 无序	✅ 可变	 ❌ 不允许	数学集合操作（交集、并集），自动去重。		

```python
from typing import List

# 初始化
my_list = [10, 20, 30]

# --- 增 (Add) ---
my_list.append(40)        # 在末尾添加: [10, 20, 30, 40]
my_list.extend([50, 60])  # 合并另一个列表: [10, 20, 30, 40, 50, 60]
my_list.insert(1, 15)     # 在索引1插入: [10, 15, 20...]

# --- 删 (Remove) ---
x = my_list.pop()         # 移除并返回最后一个元素 (60)
y = my_list.pop(0)        # 移除并返回索引0的元素 (10) - *注意：头部操作慢*
my_list.remove(20)        # 移除第一个值为20的元素 (若不存在会报错)

# --- 查/改 (Access/Modify) ---
idx = my_list.index(30)   # 查找值30的索引
count = my_list.count(30) # 统计30出现的次数
my_list.sort(reverse=True)# 倒序排序 (原地修改)
sliced = my_list[::-1]    # 切片反转 (返回新列表，不修改原列表)
```

```python
from typing import Dict

my_dict = {"name": "Gemini", "age": 1}

# --- 增/改 (Add/Update) ---
my_dict["city"] = "Tokyo" # 新增
my_dict["age"] = 2        # 修改
my_dict.update({"a": 1, "b": 2}) # 批量更新/合并

# --- 查 (Query) ---
val1 = my_dict.get("name")      # 获取值: 'Gemini'
val2 = my_dict.get("job", None) # 安全获取: 键不存在时返回 None，不报错

# --- 删 (Delete) ---
val = my_dict.pop("age")        # 删除并返回值: 2
del my_dict["city"]             # 仅删除

# --- 遍历 (Iterate) ---
keys = list(my_dict.keys())     # ['name', 'a', 'b']
items = list(my_dict.items())   # [('name', 'Gemini'), ...]
```

```python
from typing import Set

s1 = {1, 2, 3}
s2 = {3, 4, 5}

# --- 增/删 (Modify) ---
s1.add(4)               # 添加元素: {1, 2, 3, 4}
s1.remove(1)            # 删除 (不存在会报错)
s1.discard(99)          # 安全删除 (不存在也不会报错) !!! 推荐

# --- 数学运算 (Math) ---
inter = s1 & s2         # 交集 (AND): {3, 4}
union = s1 | s2         # 并集 (OR): {2, 3, 4, 5}
diff  = s1 - s2         # 差集 (只在s1不在s2): {2}
sym_diff = s1 ^ s2      # 对称差集 (不共有的): {2, 5}
```

```python
from collections import Counter

data = ['apple', 'banana', 'apple', 'orange', 'apple']
c = Counter(data)

print(c)                # Counter({'apple': 3, 'banana': 1, 'orange': 1})
print(c['apple'])       # 获取次数: 3
print(c.most_common(2)) # 获取出现最多的前2名: [('apple', 3), ('banana', 1)]

# Counter 之间可以直接加减
c2 = Counter(['apple', 'apple'])
c.subtract(c2)          # c['apple'] 变为 1
```

```python
from collections import deque

dq = deque([1, 2, 3])

dq.append(4)            # 右侧添加
dq.appendleft(0)        # 左侧添加: [0, 1, 2, 3, 4] -> 速度 O(1)

dq.pop()                # 右侧弹出
dq.popleft()            # 左侧弹出 -> 速度 O(1)

dq.rotate(1)            # 向右轮转: [3, 0, 1, 2] (常用于跑马灯效果)
```

```python
from collections import defaultdict

# 示例：将单词按首字母分组
words = ['apple', 'banana', 'avocado', 'berry']

# 只要访问不存在的键，自动创建一个空 list
dd = defaultdict(list) 

for w in words:
    dd[w[0]].append(w) 

# 结果: {'a': ['apple', 'avocado'], 'b': ['banana', 'berry']}
# 传统 dict 写法需要先 if key not in d: d[key] = []
```

```python
import heapq

nums = [3, 1, 4, 1, 5]

# 1. 列表转堆 (原地调整)
heapq.heapify(nums)     # nums 变为 [1, 1, 4, 3, 5] (满足堆性质)

# 2. 推入元素
heapq.heappush(nums, 2) # 自动放置到合适位置

# 3. 弹出最小值
min_val = heapq.heappop(nums) # 返回 1 (永远弹出最小的)

# 4. 获取最大的3个元素 (Top-K)
largest_3 = heapq.nlargest(3, [1, 10, 5, 100, 2]) # [100, 10, 5]
```
