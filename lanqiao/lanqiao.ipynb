### 数字三角形

```python
'''
动态规划算法 
状态:dp[i][j]代表到达(i,j)点时的最大和
初始化:dp[0][0]=b[0][0]
状态转移:当j=0时.dp[i][j]=dp[i-1][j]+b[i][j]
当i=j时,dp[i][j]=dp[i-1][j-1]+b[i][j]
否则,dp[i][j]=max(dp[i-1][j-1]+b[i][j],dp[i-1][j]+b[i][j])
当N为奇数时,若要满足向左下走的次数与向右下走的次数相差不超过1,则最后一定走到第N行的最中间的数,即dp[N-1][N//2]
当N为偶数时,若要满足向左下走的次数与向右下走的次数相差不超过1,则最后一定走到第N行的最中间的两个数其中一个,即max(dp[N-1][N//2-1],dp[N-1][N//2])
'''
N=int(input())
b=[]
for i in range(N):
    a=list(map(int,input().split()))
    b.append(a)
print(b)


dp=[[0 for j in range(i+1)]for i in range(N)]

def dg(dp):
    for i in range(N):
        for j in range(i+1):
            if i==j==0:
                dp[i][j]=b[0][0]
            elif j==0:
                dp[i][j] = dp[i - 1][j] + b[i][j]
            elif i==j:
                dp[i][j] = dp[i - 1][j - 1] + b[i][j]
            else:
                dp[i][j] = max(dp[i - 1][j - 1] + b[i][j], dp[i - 1][j] + b[i][j])
dg(dp)
if (N%2!=0):
    print(dp[N - 1][N // 2])
else:
    print(max(dp[N-1][N//2-1],dp[N-1][N//2]))
    
print(dp)
```

```python
#  暴力解题
import os
import sys
import copy
sys.setrecursionlimit(1500000)
# 请在此输入您的代码
#回溯DFS+path拔旗。方格表示法为——0,0=1 0,1=1 3,3=15,行数*4+列数好像是。
# 2，1=9
n=int(input())
#遍历并不是永无止境的。他需要满了则回溯，拔后则加回。
#起
vis=[[0]*n for _ in range(n)]
lie=list(map(int,input().split()))
hang=list(map(int,input().split()))
tar=(sum(hang)+sum(lie))//2#目标步数。
dirs=[[0,1],[1,0],[-1,0],[0,-1]]
p=""
def dfs(x,y,path,bu):
  global p
  vis[x][y]=1
  hang[x]-=1
  lie[y]-=1
  if hang[x]==0:
    for i in range(x):
      if hang[i]>0:
        return
  if lie[y]==0:
    for i in range(y):
      if lie[i]>0:
        return
  if x==n-1 and y==n-1:
    if sum(hang)==0 and sum(lie)==0 and bu==tar:
      
      p=path
      
      return#符合条件啊啊啊啊啊
    else:
      return#确实走到终点了，但是此路线没有符合条件。
  for xq,yq in dirs:
    xi,yi=x+xq,y+yq
    if 0<=xi<n and 0<=yi<n and vis[xi][yi]!=1 and hang[xi]>0 and lie[yi]>0:
      dfs(xi,yi,path+"/"+str(xi*n+yi),bu+1)
      vis[xi][yi]=0#回溯，
      hang[xi]+=1
      lie[yi]+=1
dfs(0,0,"0",1)
print(p)
for i in p.split("/"):
  print(i,end=" ")
  
# 4
# 2 4 3 4
# 4 3 3 3
```

### 跑步锻炼    

```python
# start = 2000-1-1 Saturday
# end = 2020-10-1 Thursday
# 归类，模拟问题；直接离利用python标准库反解出时间
from datetime import date, timedelta

start = date(2000, 1, 1)
end   = date(2020, 10, 1)

delta = timedelta(days=1)
current = start
total = 0

while current <= end:
    # weekday(): Monday=0, Sunday=6
    if current.weekday() == 0 or current.day == 1:
        total += 2
    else:
        total += 1
    current += delta

print(total)
```

### 蛇形填数

```python
# 1 2 6 7 15 ...
# 3 5 8 14 ...
# 4 9 13 ...
# 10 12 ...
# 11 ...
# ...
# 对矩阵进行解构，数学归纳法;  d=i+j−1
n = 20
a = [[0]*n for _ in range(n)]

num = 1
for d in range(1, 2*n):  # 对角线编号 1..2n-1
    if d % 2 == 1:
        # ↙ 从上往下
        i = min(d, n)
        j = d - i + 1
        while i >= 1 and j <= n:
            a[i-1][j-1] = num
            num += 1
            i -= 1
            j += 1
    else:
        # ↗ 从下往上
        j = min(d, n)
        i = d - j + 1
        while j >= 1 and i <= n:
            a[i-1][j-1] = num
            num += 1
            i += 1
            j -= 1

# 打印结果
for row in a:
    print(*[str(x).rjust(3) for x in row])
print(a[19][19])
```
